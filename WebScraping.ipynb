```python
import bs4
import requests
import pandas as pd
```

```python
soup=bs4.BeautifulSoup(open("imdb.html").read())
print(soup.prettify()) #for indentation
```

```python
# tag and div
```

```python
movie_blocks=soup.findAll('div',{"class":"lister-item-content"})
len(movie_blocks)
```

```python
movie_blocks[0]
```

```python
m_name=movie_blocks[0].find('a')
m_name
```

```python
m_name=movie_blocks[0].find('a').text
m_name
```

```python
m_name=movie_blocks[0].find('a').get_text()
m_name
```

```python
# For Year
m_year=movie_blocks[0].find('span',{'class':'lister-item-year'}).text
m_year[1:5]
```

```python
# Rating
```

```python
m_rating=movie_blocks[0].find('div',{"class":"inline-block ratings-imdb-rating"}).text
m_rating
```

```python
m_rating.strip()
```

```python
#OR 
m_rating=movie_blocks[0].find('div',{"class":"inline-block ratings-imdb-rating"}).get('data-value')
m_rating
# .text gives content while .get() gives attribute
```

```python
m_title=[]
m_year=[]
m_rate=[]
for i in range(len(movie_blocks)):
    m_title.append(movie_blocks[i].find('a').text)
    m_year.append(movie_blocks[i].find('span',{'class':"lister-item-year"}).text[-5:-1])
    m_rate.append(movie_blocks[i].find('div',{'class':"inline-block ratings-imdb-rating"}).get('data-value'))
```

```python
print(m_title)
```

```python
print(m_year)
```

```python
print(m_rate)
```

```python
page_movie_data=[]
for i in range(len(movie_blocks)):
    movie_data={}
    movie_data['title']=m_title[i]
    movie_data['year']=m_year[i]
    movie_data['rating']=m_rate[i]
    page_movie_data.append(movie_data)
```

```python
page_movie_data
```

```python
df=pd.DataFrame(page_movie_data)
```

```python
df.to_csv('movies.csv',index=False)
```

```python
url="https://subslikescript.com/movies"
result=requests.get(url) #shld be run only once otherwise multiple responses will seem like bot access
content=result.text
soup=bs4.BeautifulSoup(content,'html.parser')
print(soup.text)
box=soup.find('article',class_='main-article')
```

```python
links=[]
href=box.findAll("a")
for i in range(len(href)):
    links.append(href[i].get("href"))
links
```

```python
root="https://subslikescript.com"
for i in range(len(links)):
    url=root+links[i]
    result=requests.get(url)
    content=result.text
    soup=bs4.BeautifulSoup(content,'html.parser')
    box=soup.find("article",class_='main-article')
    title=box.find("h1").text
    script=box.find("div",class_="full-script").text.strip()
    try:
        with open(f"{title}.txt",'w',encoding='utf-8') as f:
            f.write(script)
    except Exception as e:
        print(e)
```

```python
soup=bs4.BeautifulSoup(open("politifact.html",encoding='utf-8').read())
print(soup.prettify())
```

```python
box=soup.find("div",class_="o-listease__content")
box
```

```python
site=box.findAll("a",class_="m-statement__name")
l=[]
for i in site:
    l.append(i.text.strip())
l
```

```python
by=box.findAll("footer",class_="m-statement__footer")
person=[]
date=[]
for i in by:
    t=i.text.strip().split("•")
    person.append(t[0][3:])
    date.append(t[1])
print(person)
```

```python
date
```

```python
text=box.findAll("div",class_="m-statement__quote")
quote=[]
for i in text:
    quote.append(i.find("a").text.strip())
quote
```

```python
import pandas as pd
```

```python
pd.DataFrame({"media":l,"By":person,"Date":date,"Statement":quote})
```
