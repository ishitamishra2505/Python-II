# NETWORK x

```python
import networkx as nx
import matplotlib.pyplot as plt # to display for .show()
```

```python
g=nx.Graph() # canvas creation
```

```python
g.add_node(1)
g.add_node(2)
g.add_edge(1,2)
```

```python
nx.draw_networkx(g) # ain't cartesian coordinate system 
```

```python
g.add_nodes_from(range(1,6))
g.add_edges_from([(2,3),(3,5),(3,4),(5,5)])
nx.draw_networkx(g)
```

```python
g=nx.DiGraph()
g.add_nodes_from(range(1,6))
g.add_edges_from([(2,3),(3,5),(3,4),(5,5)])
nx.draw_networkx(g)
```

```python
import networkx as nx
```

```python
g=nx.DiGraph()
g.add_nodes_from(range(1,6))
l=[[0,1,1,1,0],[1,0,0,0,1],[1,1,0,1,1],[0,0,1,0,1],[1,0,0,1,1],]
for i in range(5):
    for j in range(5):
        if l[i][j]==1:
             g.add_edge(i+1,j+1)
nx.draw_networkx(g)
```

# PLOTLY

```python
import plotly.express as px
import plotly.io as pio
import pandas as pd
```

```python
#BAR,SCATTER,LINE,PIE --interactive graphs unlike matplotlib
```

```python
df=pd.DataFrame({"Department":['IT','IT','HR','HR','Finance','Finance'],
               "Gender":['M','F','M','F','M','F'],
               "Salary":[50000,55000,45000,48000,70000,72000],
               "Experience":[3,4,2,3,5,6]})
df
```

```python
pio.renderers.default='notebook' #default is browser. To keep the chart below cell option:[html,png]
```

```python
fig=px.bar(df,x='Department',y='Salary',color='Gender',barmode='stack',hover_data=["Experience"],
          orientation='h',color_discrete_map={'M':'pink','F':'yellow'})
#In orientation, x & y need to be changed manually
fig.show()
```

```python
fig=px.bar(df,x='Department',y='Salary',color='Gender',barmode='relative')
fig.show()
```

```python
df
```

```python
#Scatter
fig=px.scatter(df,x="Experience",y="Salary",size="Salary",color="Gender",hover_data="Department",
              symbol="Gender",facet_row="Gender",facet_col="Department")
fig.show()
```

```python
fig=px.scatter_3d(df,x="Experience",y="Salary",z="Department",size="Salary",color="Gender",hover_data="Department",
              symbol="Gender")
fig.show()
```

```python
df
```

```python
fig=px.line(df,x='Experience',y='Salary',markers=True,color="Gender")
fig.show()
```

```python
fig=px.line(df,x='Experience',y='Salary',markers=True)
fig.show()
```

```python
fig=px.pie(df,values="Salary",names="Department",color_discrete_sequence=['pink','yellow','magenta'])
fig.update_traces(pull=[0,0,1],textinfo='percent+label')
fig.show()
```

```python
#HISTOGRAM ==to know the count,range
fig=px.histogram(df,x="Salary",nbins=4,color="Gender",text_auto=True)
fig.show()
```

# DASH

```python
import dash 
from dash import html,dcc
import plotly.express as px 
import pandas as pd
```

```python
data=[{"player":['Kohli','Rohit','Dhoni'],'runs':[75,60,50]}]
df=pd.DataFrame(data)
fig=px.bar(df,x='player',y='runs')
app=dash.Dash(__name__)
app.layout=html.Div([html.H1('simple dashboard'),dcc.Graph(figure=fig)])
if __name__=="__main__":
    app.run(port=8051)
```
