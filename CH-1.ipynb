```python
# marginal_error
# factor of safety=fos i.e high and low 
```

```python
low=Q1-(1.5*IQR)
print(low)
```

```python
# hence the rows having values > than high and < than low are not of any use to us
```

```python
# finding outliers
d.loc[(d['acceleration']<=low) | (d['acceleration']>=high)]
```

```python
# remove outliers
d.loc[(d['acceleration']>low) & (d['acceleration']<high)]
```

```python
df1=pd.DataFrame({'Roll':[1,2,3],'Name':['A','B','C']})
df2=pd.DataFrame({'Roll':[1,2,4],'Marks':[24,20,22]})
#df2=pd.DataFrame({'Roll_no':[1,2,4],'Marks':[24,20,22]})
```

```python
# merge :- kind of sequential joining through mapping
#pd.merge(df1,df2,on='Roll')
#pd.merge(df1,df2,on='Name') # no common col name
#pd.merge(df1,df2) # col name diff but data type same then also error
pd.merge(df1,df2,left_on='Roll',right_on='Roll_no')
```

```python
pd.merge(df1,df2,on='Roll',how='inner')
```

```python
pd.merge(df1,df2,on='Roll',how='left')
```

```python
pd.merge(df1,df2,on='Roll',how='right')
```

```python
pd.merge(df1,df2,on='Roll',how='outer')
```

```python
# concate
pd.concat((df1,df2),axis=1)  # kind of stacking def val of axis=0
```

```python
# insert a new col performance where if marks =25 performance=good
# 22-25 per=avg
# <22  perf=below avg
```

```python
def tracker(mrks):
    if mrks==25 :
        return 'good'
    elif mrks>=22 and mrks<25 :
        return 'average'
    else :
        return 'below average'

```

```python
# APPLY
df2['Performance']=df2['Marks'].apply(tracker)
```

```python
import pandas as pd
df=pd.DataFrame({'Name':['A','B','C','D'], 'Branch':['IT','AIML','IT','CSE-AI'],'Marks':[80,90,60,70]})
df.groupby('Branch') # main keyword of this mthd is 'by'
```

```python
print(df)
```

```python
df.groupby('Branch').count()
```

```python
df.groupby('Branch').sum()
```

```python
df.groupby(by='Branch',as_index=False).sum() #def value of as_index is true
```

```python
#df.groupby(by='Branch',as_index=False,sort='').sum() #def value of as_index is true
df.groupby(by='Branch',as_index=False,sort=True).sum() #def value of as_index is true
```
